```python
# ================================================================
# 🧠 HIDDENRHYTHM AI
# Behavioral Pattern Discovery & Digital Life Analytics Engine
# ONE-CELL DATA ANALYSIS + ML + VISUALIZATION PROJECT
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.ensemble import IsolationForest
from sklearn.metrics import silhouette_score

warnings.filterwarnings("ignore")
np.random.seed(42)

# ------------------------------------------------
# 1. GENERATE 365 DAYS OF BEHAVIORAL DATA
# ------------------------------------------------

n = 365

dates = pd.date_range("2025-01-01", periods=n, freq="D")

df = pd.DataFrame({
    "Date": dates,
    "Screen_Time": np.clip(np.random.normal(6.5, 1.8, n), 1, 13),
    "Sleep_Hours": np.clip(np.random.normal(7.0, 1.1, n), 3.5, 10),
    "Study_Hours": np.clip(np.random.normal(3.5, 1.5, n), 0, 9),
    "Social_Media": np.clip(np.random.normal(2.2, 1.0, n), 0, 7),
    "Coding_Hours": np.clip(np.random.normal(2.0, 1.3, n), 0, 7),
    "Notifications": np.random.poisson(85, n),
    "App_Switches": np.random.poisson(110, n),
    "Exercise_Min": np.clip(np.random.normal(35, 20, n), 0, 120),
})

# ------------------------------------------------
# 2. CREATE REALISTIC RELATIONSHIPS
# ------------------------------------------------

# More social media tends to increase screen time
df["Screen_Time"] += df["Social_Media"] * 0.35

# Heavy screen use slightly reduces sleep
df["Sleep_Hours"] -= np.maximum(df["Screen_Time"] - 7, 0) * 0.12

# Weekend behavior
df["Day"] = df["Date"].dt.day_name()
df["Month"] = df["Date"].dt.month_name().str[:3]
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)
df["Is_Weekend"] = df["Date"].dt.dayofweek >= 5

df.loc[df["Is_Weekend"], "Social_Media"] *= 1.25
df.loc[df["Is_Weekend"], "Study_Hours"] *= 0.72
df.loc[df["Is_Weekend"], "Sleep_Hours"] += 0.45

# Introduce several intense workload periods
for start, end in [(65, 82), (170, 190), (285, 305)]:
    df.loc[start:end, "Study_Hours"] *= 1.45
    df.loc[start:end, "Coding_Hours"] *= 1.35
    df.loc[start:end, "Sleep_Hours"] *= 0.82
    df.loc[start:end, "Screen_Time"] *= 1.18
    df.loc[start:end, "Notifications"] *= 1.25

# Keep values realistic
df["Sleep_Hours"] = df["Sleep_Hours"].clip(3, 10)
df["Screen_Time"] = df["Screen_Time"].clip(1, 15)
df["Study_Hours"] = df["Study_Hours"].clip(0, 10)
df["Coding_Hours"] = df["Coding_Hours"].clip(0, 8)

# ------------------------------------------------
# 3. FEATURE ENGINEERING
# ------------------------------------------------

df["Deep_Work"] = df["Study_Hours"] + df["Coding_Hours"]

df["Focus_Fragmentation"] = (
    df["App_Switches"] / 120 +
    df["Notifications"] / 100 +
    df["Social_Media"] / 3
)

df["Digital_Load"] = (
    df["Screen_Time"] +
    df["Notifications"] / 40 +
    df["App_Switches"] / 60
)

df["Recovery_Score"] = (
    df["Sleep_Hours"] * 8 +
    df["Exercise_Min"] * 0.15 -
    df["Screen_Time"] * 2
)

df["Productivity_Score"] = (
    df["Study_Hours"] * 6 +
    df["Coding_Hours"] * 7 +
    df["Sleep_Hours"] * 2 -
    df["Social_Media"] * 3 -
    df["Focus_Fragmentation"] * 2
)

# Project-defined workload strain indicator
raw_risk = (
    df["Screen_Time"] * 5 +
    df["Social_Media"] * 5 +
    df["Notifications"] * 0.07 +
    df["App_Switches"] * 0.05 +
    np.maximum(7 - df["Sleep_Hours"], 0) * 10 +
    df["Deep_Work"] * 2 -
    df["Exercise_Min"] * 0.08
)

df["Workload_Strain_Index"] = np.clip(raw_risk, 0, 100)

# ------------------------------------------------
# 4. UNSUPERVISED ML — FIND HIDDEN BEHAVIOR TYPES
# ------------------------------------------------

features = [
    "Screen_Time",
    "Sleep_Hours",
    "Study_Hours",
    "Social_Media",
    "Coding_Hours",
    "Notifications",
    "App_Switches",
    "Exercise_Min"
]

X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Automatically evaluate K = 2..6
silhouette_scores = {}

for k in range(2, 7):
    temp_model = KMeans(n_clusters=k, random_state=42, n_init=20)
    labels = temp_model.fit_predict(X_scaled)
    silhouette_scores[k] = silhouette_score(X_scaled, labels)

best_k = max(silhouette_scores, key=silhouette_scores.get)

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=30
)

df["Cluster"] = kmeans.fit_predict(X_scaled)

# ------------------------------------------------
# 5. NAME CLUSTERS FROM THEIR CHARACTERISTICS
# ------------------------------------------------

cluster_profile = df.groupby("Cluster")[[
    "Deep_Work",
    "Social_Media",
    "Sleep_Hours",
    "Screen_Time",
    "Focus_Fragmentation",
    "Workload_Strain_Index"
]].mean()

names = {}

for c, row in cluster_profile.iterrows():

    if row["Deep_Work"] > cluster_profile["Deep_Work"].median() and \
       row["Workload_Strain_Index"] > cluster_profile["Workload_Strain_Index"].median():
        name = "Intense Achiever"

    elif row["Social_Media"] > cluster_profile["Social_Media"].median() and \
         row["Focus_Fragmentation"] > cluster_profile["Focus_Fragmentation"].median():
        name = "Digitally Fragmented"

    elif row["Sleep_Hours"] > cluster_profile["Sleep_Hours"].median() and \
         row["Workload_Strain_Index"] < cluster_profile["Workload_Strain_Index"].median():
        name = "Balanced Recovery"

    elif row["Deep_Work"] > cluster_profile["Deep_Work"].median():
        name = "Focused Builder"

    else:
        name = "Low-Intensity Day"

    # Avoid duplicate names
    if name in names.values():
        name += f" {c}"

    names[c] = name

df["Behavior_Type"] = df["Cluster"].map(names)

# ------------------------------------------------
# 6. PCA — CREATE 2D BEHAVIOR MAP
# ------------------------------------------------

pca = PCA(n_components=2)

coords = pca.fit_transform(X_scaled)

df["PCA1"] = coords[:, 0]
df["PCA2"] = coords[:, 1]

# ------------------------------------------------
# 7. ANOMALY DETECTION
# ------------------------------------------------

iso = IsolationForest(
    contamination=0.045,
    random_state=42
)

df["Anomaly"] = iso.fit_predict(X_scaled)

df["Unusual_Day"] = df["Anomaly"] == -1

# ------------------------------------------------
# 8. BEHAVIOR TRANSITION DETECTION
# ------------------------------------------------

df["Previous_Behavior"] = df["Behavior_Type"].shift(1)

df["Behavior_Transition"] = (
    df["Behavior_Type"] != df["Previous_Behavior"]
)

df.loc[0, "Behavior_Transition"] = False

# ------------------------------------------------
# 9. CORRELATION ANALYSIS
# ------------------------------------------------

analysis_cols = [
    "Screen_Time",
    "Sleep_Hours",
    "Study_Hours",
    "Social_Media",
    "Coding_Hours",
    "Notifications",
    "App_Switches",
    "Exercise_Min",
    "Deep_Work",
    "Productivity_Score",
    "Workload_Strain_Index"
]

corr = df[analysis_cols].corr()

# ------------------------------------------------
# 10. VISUALIZATION DASHBOARD
# ------------------------------------------------

fig = plt.figure(figsize=(18, 24))

fig.suptitle(
    "HIDDENRHYTHM AI — Behavioral Intelligence Analysis",
    fontsize=24,
    fontweight="bold",
    y=0.995
)

# -------- CHART 1: PCA BEHAVIOR MAP --------

ax1 = plt.subplot(3, 2, 1)

for behavior in df["Behavior_Type"].unique():

    temp = df[df["Behavior_Type"] == behavior]

    ax1.scatter(
        temp["PCA1"],
        temp["PCA2"],
        alpha=0.65,
        label=behavior
    )

ax1.set_title("Hidden Behavioral Archetypes", fontweight="bold")
ax1.set_xlabel("Behavior Dimension 1")
ax1.set_ylabel("Behavior Dimension 2")
ax1.legend(fontsize=8)


# -------- CHART 2: WORKLOAD STRAIN TIMELINE --------

ax2 = plt.subplot(3, 2, 2)

ax2.plot(
    df["Date"],
    df["Workload_Strain_Index"],
    linewidth=1.5
)

ax2.axhline(
    df["Workload_Strain_Index"].quantile(.80),
    linestyle="--",
    label="Top 20% strain threshold"
)

ax2.set_title("Workload Strain Timeline", fontweight="bold")
ax2.set_ylabel("Strain Index")
ax2.legend()
ax2.tick_params(axis="x", rotation=45)


# -------- CHART 3: SCREEN VS PRODUCTIVITY --------

ax3 = plt.subplot(3, 2, 3)

scatter = ax3.scatter(
    df["Screen_Time"],
    df["Productivity_Score"],
    c=df["Workload_Strain_Index"],
    alpha=.7
)

ax3.set_title(
    "Screen Time vs Productivity",
    fontweight="bold"
)

ax3.set_xlabel("Screen Time (hours)")
ax3.set_ylabel("Productivity Score")

plt.colorbar(
    scatter,
    ax=ax3,
    label="Workload Strain"
)


# -------- CHART 4: CORRELATION MATRIX --------

ax4 = plt.subplot(3, 2, 4)

im = ax4.imshow(
    corr,
    aspect="auto"
)

ax4.set_xticks(range(len(corr.columns)))
ax4.set_yticks(range(len(corr.columns)))

ax4.set_xticklabels(
    corr.columns,
    rotation=90,
    fontsize=8
)

ax4.set_yticklabels(
    corr.columns,
    fontsize=8
)

ax4.set_title(
    "Behavior Correlation Matrix",
    fontweight="bold"
)

plt.colorbar(im, ax=ax4)


# -------- CHART 5: WEEKLY DIGITAL LOAD --------

ax5 = plt.subplot(3, 2, 5)

weekly = df.groupby("Week")[
    ["Digital_Load", "Deep_Work"]
].mean()

ax5.plot(
    weekly.index,
    weekly["Digital_Load"],
    label="Digital Load"
)

ax5.plot(
    weekly.index,
    weekly["Deep_Work"],
    label="Deep Work"
)

ax5.set_title(
    "Digital Load vs Deep Work by Week",
    fontweight="bold"
)

ax5.set_xlabel("Week")
ax5.legend()


# -------- CHART 6: BEHAVIOR TYPE FREQUENCY --------

ax6 = plt.subplot(3, 2, 6)

counts = df["Behavior_Type"].value_counts()

ax6.barh(
    counts.index,
    counts.values
)

ax6.set_title(
    "Dominant Behavioral Patterns",
    fontweight="bold"
)

ax6.set_xlabel("Number of Days")

plt.tight_layout(rect=[0, 0, 1, .98])
plt.show()

# ------------------------------------------------
# 11. AUTOMATIC ANALYTICAL REPORT
# ------------------------------------------------

highest_strain_day = df.loc[
    df["Workload_Strain_Index"].idxmax()
]

most_productive_day = df.loc[
    df["Productivity_Score"].idxmax()
]

dominant_behavior = df["Behavior_Type"].mode()[0]

anomaly_count = df["Unusual_Day"].sum()

transition_count = df["Behavior_Transition"].sum()

screen_productivity_corr = df[
    ["Screen_Time", "Productivity_Score"]
].corr().iloc[0, 1]

sleep_strain_corr = df[
    ["Sleep_Hours", "Workload_Strain_Index"]
].corr().iloc[0, 1]

# ------------------------------------------------
# 12. PRINT REPORT
# ------------------------------------------------

print("\n" + "=" * 72)
print("🧠 HIDDENRHYTHM AI — AUTOMATIC INTELLIGENCE REPORT")
print("=" * 72)

print(f"""
DATASET
-------
Days analyzed               : {len(df)}
Detected behavior types      : {best_k}
Unusual behavioral days      : {anomaly_count}
Behavior transitions         : {transition_count}

DOMINANT BEHAVIOR
-----------------
{dominant_behavior}

MOST PRODUCTIVE DAY
-------------------
Date                         : {most_productive_day['Date'].date()}
Productivity Score           : {most_productive_day['Productivity_Score']:.2f}
Deep Work                    : {most_productive_day['Deep_Work']:.2f} hours
Sleep                        : {most_productive_day['Sleep_Hours']:.2f} hours

HIGHEST STRAIN DAY
------------------
Date                         : {highest_strain_day['Date'].date()}
Workload Strain Index        : {highest_strain_day['Workload_Strain_Index']:.2f}
Screen Time                  : {highest_strain_day['Screen_Time']:.2f} hours
Sleep                        : {highest_strain_day['Sleep_Hours']:.2f} hours

RELATIONSHIPS
-------------
Screen Time ↔ Productivity   : {screen_productivity_corr:.3f}
Sleep ↔ Workload Strain      : {sleep_strain_corr:.3f}

ML QUALITY
----------
Best K                       : {best_k}
Silhouette Score             : {silhouette_scores[best_k]:.3f}

PCA Variance Explained       : {pca.explained_variance_ratio_.sum()*100:.2f}%
""")

print("DISCOVERED BEHAVIOR TYPES")
print("-" * 72)

for cluster, name in names.items():

    subset = df[df["Cluster"] == cluster]

    print(
        f"{name:25s} | "
        f"{len(subset):3d} days | "
        f"Deep Work {subset['Deep_Work'].mean():.1f}h | "
        f"Sleep {subset['Sleep_Hours'].mean():.1f}h | "
        f"Strain {subset['Workload_Strain_Index'].mean():.1f}"
    )

print("\nTOP 5 MOST UNUSUAL DAYS")
print("-" * 72)

print(
    df[df["Unusual_Day"]][
        [
            "Date",
            "Behavior_Type",
            "Screen_Time",
            "Sleep_Hours",
            "Deep_Work",
            "Workload_Strain_Index"
        ]
    ]
    .sort_values(
        "Workload_Strain_Index",
        ascending=False
    )
    .head()
    .to_string(index=False)
)

print("\n" + "=" * 72)
print("Analysis complete.")
print("Workload Strain Index is a project-defined analytical metric,")
print("not a medical or psychological diagnosis.")
print("=" * 72)
```
